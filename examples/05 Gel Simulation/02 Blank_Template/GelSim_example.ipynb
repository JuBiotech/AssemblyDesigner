# digest_and_plot_plasmids

Simulates restriction enzyme digests on plasmid DNA sequences and visualizes the resulting fragment sizes on a simulated agarose gel.

## Description

This function takes a list of plasmid files together with specified restriction enzymes for each plasmid, performs in-silico restriction digests, and plots a gel showing the fragment sizes for each sample alongside a DNA ladder.

The plasmid files can be in GenBank (`.gb`) or SnapGene (`.dna`) formats. The DNA ladder type is customizable, allowing visualization with common ladder standards such as "10kb", "5kb", "2kb", "1kb", or "100bp".

## Parameters

- `plasmid_files_with_enzymes`: List of tuples `(file_path, enzymes)`  
  Each tuple contains the path to a plasmid file and a list of enzyme names to use for the digest.

- `ladder_type`: `str`, optional (default: `"2kb"`)  
  Specifies the type of DNA ladder to use in the gel simulation. Supported values include `"10kb"`, `"5kb"`, `"2kb"`, `"1kb"`, and `"100bp"`.

## Example

```python
plasmid_files_with_enzymes = [
    ("C0062_CD.gb", ["EcoRI", "BamHI", "BpmI"]),
    ("C0040_CD.gb", ["HindIII", "BamHI", "BsaI"]),
    ("C0012m_CD.gb", ["EcoRI", "XbaI", "BsaI"]),
    ("J23100_B0032_GFP_B0015_Assembled.dna", ["EcoRI", "BpmI"]),
]

digest_and_plot_plasmids(plasmid_files_with_enzymes, ladder_type="2kb")
